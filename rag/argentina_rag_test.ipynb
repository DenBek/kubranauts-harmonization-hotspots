## 3. Query

```python
def ask(q, k=5):
    print('Q:', q)
    for h in rag.query(q, k):
        print(f"  [{h['score']:.3f}] {h['file']}  {h['heading']}")
        print('     ', ' '.join(h['text'][:300].split()))

ask('why does VIIRS count more fires than MODIS in Argentina')
```
